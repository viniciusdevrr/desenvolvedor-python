```python
import numpy as np

renda = np.array([5000, 4500, 6000, 4000])
gasto = np.array([4000, 3500, 6000, 3900])

lucro = renda - gasto

print(f"Lucro dos meses: {lucro}")
```

```python
notas_alunos = np.array([
    10.0,6.0,5.6,7.4,8.8
])

# notas_aprovados = notas_alunos[~(notas_alunos >= 7)]
notas_aprovados = notas_alunos[notas_alunos >= 7]
print(f"Notas aprovadas: {notas_aprovados}")

notas_reprovados = notas_alunos[notas_alunos < 7]
print(f"Notas reprovadas: {notas_reprovados}")
```

```python
vendas = np.array([
    [15000, 20000, 30000], # 1º Trimestre
    [25000, 30000, 40000], # 2º Trimestre
    [35000, 40000, 50000], # 3º Trimestre
    [45000, 50000, 60000], # 4º Trimestre
])

total_vendas_ano = sum(vendas)
print(f"Total de vendas: {total_vendas_ano}\n")

total_por_trimestre = np.sum(vendas, axis=1)
print(f"Calcula o valor de cada linha do Array")
print(f"Total de venda por trimestre: {total_por_trimestre}\n")

comeco_de_trimestre = np.sum(vendas, axis=0)
print(f"Calcula o valor total de cada coluna")
print(f"Total comeco do trimestre: {comeco_de_trimestre}\n")

maior_venda_ano = np.max(vendas)
print(f"Maior venda ano: {maior_venda_ano}\n")

menor_venda_ano = np.min(vendas)
print(f"Menor venda ano: {menor_venda_ano}")
```

```python
valor_produtos = np.array([
    10,15,20,30
])

valor_promocao = np.where(
    valor_produtos >= 20, valor_produtos * 0.9, valor_produtos
)

print(valor_promocao)
```
