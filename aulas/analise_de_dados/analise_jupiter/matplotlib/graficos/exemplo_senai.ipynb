```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.core.pylabtools import figsize

df_aluno = pd.read_csv('alunos_senai_sig.csv')
```

```python
contagem_de_cursos = df_aluno.groupby('Curso').size()
print(contagem_de_cursos)

# passa a base de dados criada (contagem dos cursos)
# plt.pie = Grafico de Pizza
# plt.pie acompanha ->
plt.pie(contagem_de_cursos,
        labels=contagem_de_cursos.index, # Mostrar Nomes dos Dados
        autopct='%1.1f%%', # Exibir Valores em Porcetagem com CS Decimais
        colors=['b', 'r', 'g', 'y'],) # Exibir cores (qtd) referente aos dados

plt.title('Cursos de Senai')
plt.show()
```

GRAFICO DE MÉDIA DE CADA TURMA

```python
media_por_curso = df_aluno.groupby('Curso')['Media_Final'].mean() .sort_values(ascending=False)

# plt.figure(figsize=(largura , altura) da figura
plt.figure(figsize=(8,4))

# plt.bar = Grafico de Barra
# plt.pie = Grafico de Pizza
barra = plt.bar(
    media_por_curso.index,
    media_por_curso
)

plt.title("Turma")
plt.ylabel("Média por Curso")
plt.xlabel("Cursos")

plt.ylim(0, 11)

for coluna in barra:
    yvalor = coluna.get_height()
    plt.text(
        coluna.get_x() + coluna.get_width() / 2,
        yvalor,
        f"{yvalor:.2f}",
        ha='center',
        va='bottom',
        fontsize=10
    )


plt.show()
```

LIDANDO COM SUBPLOT

```python
fig, axs = plt.subplots(2,2, figsize=(20,10))

df_python = df_aluno[df_aluno['Curso'] == 'Python']
df_java = df_aluno[df_aluno['Curso'] == 'Java']
df_banco_de_dados = df_aluno[df_aluno['Curso'] == 'Banco de Dados']
df_design = df_aluno[df_aluno['Curso'] == 'Design']

axs[0,0].barh(df_python["Nome"],  df_python['Media_Final'])
axs[0,0].set_title("Python")

axs[0,1].plot(df_java["Nome"], df_java['Media_Final'])
axs[0,1].set_title("Java")

axs[1,0].scatter(df_banco_de_dados['Media_Final'], df_banco_de_dados["Nome"])
axs[1,0].set_title("Banco de Dados")

axs[1,1].scatter(df_design['Media_Final'], df_design["Nome"])
axs[1,1].set_title("Design")

plt.show()
```
