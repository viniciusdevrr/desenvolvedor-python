```python
import pandas as pd
import matplotlib.pyplot as plt

dados_turma = [
    {
        'nome': 'Vinicius',
        'idade': 22,
        'materia': 'Python',
        'nota_final': 9
    },
    {
        'nome': 'Dante',
        'idade': 33,
        'materia': 'Java',
        'nota_final': 7
    },
    {
        'nome': 'Neto',
        'idade': 26,
        'materia': 'Ruby',
        'nota_final': 2
    },
    {
        'nome': 'Jarbas',
        'idade': 35,
        'materia': 'Java',
        'nota_final': 8
    },
    {
        'nome': 'Indaia',
        'idade': 26,
        'materia': 'Ruby',
        'nota_final': 7
    },
    {
        'nome': 'Mauro',
        'idade': 26,
        'materia': 'Python',
        'nota_final': 4
    }
]
df_turmas = pd.DataFrame(dados_turma)

media_turmas = df_turmas.groupby(['materia'])['nota_final'].mean()

print(media_turmas)
```

```python
plt.plot(
    media_turmas.max(),
    media_turmas.min(),
    color='red'
)

plt.xlabel("Turmas")
plt.xlabel("Nota Final")


plt.grid(True, linestyle=':')
plt.show()
```
