```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df_aluno_senai = pd.read_csv('alunos_senai_sig.csv')
print(df_aluno_senai)
```

```python
media_por_curso = (
    df_aluno_senai.groupby('Curso')['Media_Final'].mean().sort_values(ascending=False)
)

plt.figure(figsize=(8, 4))

barra = plt.bar(media_por_curso.index, media_por_curso, color='skyblue')

plt.title('Média por Curso')
plt.ylabel('Média Final')
plt.xlabel('Cursos')

plt.ylim(0, 11)

for coluna in barra:
  yvalor = coluna.get_height()
  plt.text(
      coluna.get_x() + coluna.get_width() / 2,
      yvalor,
      f'{yvalor:.2f}',
      ha='center',
      va='bottom',
      fontsize=10,
  )

plt.show()
```

```python
# Filtrando os cursos
df_design = df_aluno_senai[df_aluno_senai['Curso'] == 'Design'].reset_index(drop=True)
df_java = df_aluno_senai[df_aluno_senai['Curso'] == 'Java'].reset_index(drop=True)

plt.figure(figsize=(10, 5))

plt.plot(df_design['Media_Final'], marker='o')
plt.plot(df_java['Media_Final'], marker='s')

plt.text(7, 9.9, 'maior nota', color='red')  # No Design
plt.text(5, 10.0, 'maior nota', color='red')  # No Java

plt.title('Gráfico PLOT - Design e Java')
plt.xlabel('Alunos')
plt.ylabel('Média Final')
plt.show()
```
