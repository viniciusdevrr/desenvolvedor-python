```python
import pandas as pd
import numpy as np

df = pd.read_csv("vendas_mercados.csv")
```

```python
df["Lucro"] = (
    (df["Preco_Unitario"] - df["Custo_Unitario"]) *
    df["Quantidade"]
)

lucro_por_mercado = df.groupby("Mercado")["Lucro"].sum()

print("\n--- LUCRO TOTAL POR MERCADO ---")
print(lucro_por_mercado)

maior_valor = 0
maior_mercado = ""

for mercado, lucro in lucro_por_mercado.items():

    if lucro > maior_valor:
        maior_valor = lucro
        maior_mercado = mercado

print("\nMercado com maior lucro:", maior_mercado)
print(f"Lucro total: R$ {maior_valor:.2f}")
```

```python
for regiao in ["Centro", "Sul", "Norte", "Leste", "Oeste"]:

    dados_regiao = df[df["Regiao"] == regiao]

    pagamentos = dados_regiao.groupby("Forma_Pagamento")["ID_Venda"].count()

    print(f"\nRegião: {regiao}")
    print(pagamentos)
```

```python
vendas_por_produto = df.groupby("Produto")["Quantidade"].sum()

print("\n--- QUANTIDADE VENDIDA POR PRODUTO ---")
print(vendas_por_produto)

maior_quantidade = 0
menor_quantidade = 999999

produto_mais_vendido = ""
produto_menos_vendido = ""

for produto, quantidade in vendas_por_produto.items():

    if quantidade > maior_quantidade:
        maior_quantidade = quantidade
        produto_mais_vendido = produto

    if quantidade < menor_quantidade:
        menor_quantidade = quantidade
        produto_menos_vendido = produto

print(f"\nProduto mais vendido: {produto_mais_vendido}")
print(f"Quantidade: {maior_quantidade}")

print(f"\nProduto menos vendido: {produto_menos_vendido}")
print(f"Quantidade: {menor_quantidade}")
```

```python
df["Faturamento"] = (
    df["Preco_Unitario"] * df["Quantidade"]
)

faturamento_total = df["Faturamento"].sum()

faturamento_por_mercado = df.groupby("Mercado")["Faturamento"].sum()

print("\n--- FATURAMENTO POR MERCADO ---")
print(faturamento_por_mercado)

print(f"\nFaturamento total da rede: R$ {faturamento_total:.2f}")

maior_receita = 0
mercado_maior_receita = ""

for mercado, valor in faturamento_por_mercado.items():

    if valor > maior_receita:
        maior_receita = valor
        mercado_maior_receita = mercado

print(f"\nMercado com maior receita: {mercado_maior_receita}")
print(f"Faturamento: R$ {maior_receita:.2f}")
```

```python
# 6
df["Preco_Promocional"] = np.where(
    df["Quantidade"] >= 10,
    df["Preco_Unitario"] * 0.90,
    df["Preco_Unitario"]
)

df["Faturamento_Promocional"] = (
    df["Preco_Promocional"] * df["Quantidade"]
)

novo_faturamento = df["Faturamento_Promocional"].sum()

print("\n--- SIMULAÇÃO DE PROMOÇÃO ---")

print(f"Faturamento original: R$ {faturamento_total:.2f}")

print(f"Novo faturamento com promoção: R$ {novo_faturamento:.2f}")

"""
Explicação e justificativa da resposta:

Utilizei np.where() para criar uma nova coluna chamada Preco_Promocional. Quando a quantidade comprada é maior ou igual a 10 unidades, o cliente recebe 10% de desconto no preço unitário. Caso contrário, o preço permanece igual.

Depois, criei a coluna Faturamento_Promocional, multiplicando o preço promocional pela quantidade comprada. Por fim, utilizei sum() para calcular o novo faturamento total da rede.

Com a aplicação da promoção, o faturamento da rede seria de R$ 4.528,31.
"""
```
